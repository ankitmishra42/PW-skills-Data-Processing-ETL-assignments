## Data Quality and Validation

**Q1: Define Data Quality in the context of ETL pipelines. Why is it more than just data cleaning?**
- Data Quality is the measure of how data is for business decisions. It is measured by dimensions like Accuracy, Completeness, Consistency, Validity, and Timeliness.

- **Why it is more than just Data Cleaning:** *Data cleaning is a subset of Data Quality. The key differences are like these.*

  - **Reactive vs Proactive:** *Cleaning fixes errors after they exist (e.g., removing nulls), Quality involves setting up "Data Contracts" and validation gates to prevent errors from entering the pipeline.*

  - **Structural vs Contextual:** *Cleaning focuses on formatting (e.g., fixing a date string). Quality focuses on business logic (e.g., ensuring a "clean" price of $10.00 makes sense for a high-value item).*

  - **Pipeline Health:** *Data Quality includes Observability monitoring data volume, freshness, and schema changes which goes far beyond simply editing values.*
  
  - **Root Cause:** *Cleaning treats symptoms; Data Quality uses lineage to identify and fix the source of the problem.*

**Q2: Explain why poor data quality leads to misleading dashboards and incorrect decisions.**
- Poor data quality leads to misleading dashboards and incorrect decisions because:
  - Incorrect or missing data produces wrong metrics, trends, and KPIs.
  - Inconsistent data formats cause aggregation errors and inaccurate comparisons.
  - Duplicate records inflate counts and distort performance results.
  - Outdated data reflects past conditions, leading to decisions that don’t match current reality.
  - Biased or incomplete data hides important patterns and gives a false sense of insight.
  - Low trust in dashboards makes stakeholders either overreact to wrong insights or ignore analytics altogether.

<br>

**Q3: What is duplicate data? Explain three causes in ETL pipelines.**
- Duplicate data means multiple records that represent the same real-world entity or event but appear more than once in a dataset.
- Three causes in ETL pipelines:

  - **Multiple ingestion paths**
    - Data arrives from different systems or files that contain overlapping records. Without proper matching rules or a master key, the ETL process treats the same entity as new each time.

  - **Faulty incremental or reprocessing logic**
    - ETL jobs may reload historical data due to missing timestamps, incorrect watermark handling, or rerunning failed jobs—resulting in repeated inserts of the same records.

  - **Weak data validation at load stage**
    - When the target system lacks primary keys, unique constraints, or deduplication checks, duplicate rows are not blocked during the load phase.

**Q4: Differentiate between exact, partial, and fuzzy duplicates.**
- Exact, Partial, and Fuzzy Duplicates Key Differences:

- **Exact duplicates**
  - Records that are completely identical across all fields.
  - Example: *Same customer ID, name, email, and phone repeated multiple times.*

- **Partial duplicates**
  - Records that match on key fields but differ in some attributes.
  - Example: *Same email ID but different address or phone number.*

- **Fuzzy duplicates**
  - Records that refer to the same entity but have spelling variations, formatting differences, or minor - errors.
  - Example: *“Ankit Mishra” vs “Ankeet Mishra” or “ankit@gmail.com
  ” vs “Ankit@gmail.com”*

**Q5: Why should data validation be performed during transformation rather than after loading?**
- Performing data validation during transformation is better because it is proactive rather than reactive.
  - **Prevents Contamination:** It keeps "garbage" out of your production tables, ensuring stakeholders never see incorrect reports.
  - **Lower Cost:** It is cheaper to fix data in the pipeline than to run expensive "delete and update" queries on a live database.
  - **Prevents Crashes:** Catching schema or constraint violations early prevents the final "Load" step from failing halfway through.
  - **Easier Debugging:** We can instantly tell if the error came from the source data or our transformation logic before it gets mixed with existing records.

<br>

**Q6: Explain how business rules help in validating data accuracy. Give an example.**
- Business rules define logical conditions based on real-world operations that data must follow, helping ensure data accuracy and reliability.

- How business rules help in validation:
  - They check whether data values make sense in a business context.
  - They prevent logically incorrect or impossible records from being processed.
  - They ensure consistency between related fields.
  - They align data with organizational policies and processes.

- **Example:**
  - In a sales system, a business rule may state that Order **Amount = Quantity × Unit Price**.
  - During data transformation, records where the order amount does not match this calculation are flagged or rejected.

- **Result:**
  - By applying business rules, inaccurate data is identified early, ensuring only valid and meaningful data is used for reporting and decision-making.
