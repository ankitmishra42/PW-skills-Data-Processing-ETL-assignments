## Data Extraction in ETL

**Q1: Describe different types of data sources used in ETL with suitable examples.**

- Different types of data sources used in ETL:
  - **Relational Databases**: Structured data stored in tables, such as MySQL, PostgreSQL, Oracle.
  - **Flat Files**: CSV, Excel, JSON, XML.
  - **APIs / Web Services**: REST APIs, from payment gateways or social media platforms.
  - **Data Warehouses**: Amazon Redshift, Snowflake, Google BigQuery.
  - **Streaming / Log Data**: Kafka streams, application server logs.
  - **Cloud Storage**: AWS S3, Azure Data Lake.

<br>

**Q2: What is data extraction? Explain its role in the ETL pipeline.**
- Data extraction is the process of collecting raw data from different source systems and bringing it into the ETL pipeline for further processing.

- Role in the ETL pipeline:
  - *It acts as the first step of ETL.*
  - *It retrieves data from multiple sources such as databases, files, APIs, or logs.*
  - *It ensures data is captured accurately and completely from source systems.*
  - *It prepares data for transformation without impacting source system performance.*
  - *It enables downstream processes like data cleaning, transformation, and loading.*

<br>

**Q3: Explain the difference between CSV and Excel in terms of extraction and ETL usage.**
- Difference between CSV and Excel in ETL extraction and usage:
  - **CSV:** Plain text, easy and fast to extract, lightweight, ideal for automated ETL and large datasets.
  - **Excel:** Supports multiple sheets, formulas, and formatting, extraction is slower, best for small or manual ETL tasks.

<br>

**Q4: Explain the steps involved in extracting data from a relational database.**
- Steps involved in extracting data from a relational database:
  - *First, we connect to the database using the required credentials.*
  - *Identify the necessary tables and columns based on the requirement.*
  - *Write and execute SQL queries to extract the data.*
  - *Apply filters or conditions to fetch only relevant records.*
  - *Validate the extracted data for accuracy and completeness.*
  - *Load the data into a staging area for further transformation.*

<br>

**Q5: Explain three common challenges faced during data extraction.**
- Three common challenges faced during data extraction:
  - Data quality issues: *Source data may contain missing, duplicate, or inconsistent values.*
  - Data source complexity: *Data may come from multiple systems, formats, or schemas.*
  - Performance impact: *Extracting large volumes of data can slow down source systems.*

<br>

**Q6: What are APIs? Explain how APIs help in real-time data extraction.**
- APIs are interfaces that allow different software systems to communicate and exchange data.
- Use of APIs in real-time data extraction:
  - *They provide direct access to live data from applications.*
  - *Data can be fetched on demand or at frequent intervals.*
  - *They support near real-time updates instead of batch loads.*
  - *They enable secure and controlled data access using authentication.*

<br>

**Q7: Why are databases preferred for enterprise-level data extraction?**
- Databases are preferred for enterprise-level data extraction because:
  - *They store large amounts of structured data efficiently.*
  - *They support SQL, so extracting data becomes flexible and reliable.*
  - *They maintain data integrity, consistency, and accuracy.*
  - *They handle enterprise workloads with high performance and scalability.*

<br>

**Q8: What steps should an ETL developer take when extracting data from large CSV files (1GB+)?**

- Steps an ETL developer should take when extracting data from large CSV files:
  - First, we don't load the whole file at once, we read it in chunks.
  - *We check the file structure and data types before extracting everything.*
  - *We apply filters while reading the file to reduce unnecessary data.*
  - *We handle memory properly so the system doesn’t crash.*
  - *We keep logs to track errors or skipped records.*
  - *Finally, we store the extracted data in a staging area for further processing.*
