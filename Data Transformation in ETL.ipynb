## Transformation in ETL

**Q1: Define Data Transformation in ETL and explain why it is important.**
- Data Transformation in ETL is the process of converting raw extracted data into a clean, structured, and meaningful format suitable for analysis or reporting.

- why it is important:
  - It cleans and standardizes data.
  - It applies business rules and calculations.
  - It improves data quality.
  - It prepares data for accurate analysis and reporting.

<br>

**Q2: List any four common activities involved in Data Cleaning.**
- Four common activities involved in Data Cleaning are like this:
  - Removing duplicate records
  - Handling missing values
  - Correcting inconsistent or invalid data
  - Standardizing data formats

<br>

**Q3: What is the difference between Normalization and Standardization?**
- Difference between Normalization and Standardization in ETL:

  - **Normalization:**
    Normalization means bringing data into a common format or scale so it stays consistent across systems.<br>
    *Example:* converting different date formats or scaling values to a common range before loading.

  - **Standardization:**
    Standardization means applying fixed rules and conventions to make data uniform.<br>
    *Example:* using the same column names, units (USD, INR), text case, or codes across datasets.

<br>

**Q4: A dataset has missing values in the “Age” column. Suggest two techniques to handle this and explain when they should be used.**
- Two ways to handle missing values in the "Age" column:

  - *Fill with mean or median age:* We will use this when only a few values are missing and age is important for analysis.
  - *Remove the records:* We will use this when very few rows have missing age or when age is not a key field.

**Q6: What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.**
- One-Hot Encoding is a technique used to convert categorical data into multiple binary columns so it can be easily used in data processing and analytics.

- **Example (Categories: Red, Blue, Green):**

| Color | Red | Blue | Green |
|-------|-----|------|-------|
| Red   | 1   | 0    | 0     |
| Blue  | 0   | 1    | 0     |
| Green | 0   | 0    | 1     |

**Q7: Explain the difference between Data Integration and Data Mapping in ETL.**
### Difference between Data Integration and Data Mapping in ETL

- **Data Integration:**  
  Combining data from different sources (databases, files, APIs) into a unified dataset for analysis or loading into a target system.<br>
  *Example:* Merging sales data from multiple regional databases into a central warehouse.

- **Data Mapping:**  
  Defining how each source field maps to the target field, including any transformations needed.<br>
  *Example:* Mapping the source column `emp_name` to the target column `EmployeeName` and converting all names to proper case.
  

<br>

**Q8: Explain why Z-score Standardization is preferred over Min-Max Scaling when outliers exist.**
- Why Z-score Standardization is preferred over Min-Max Scaling when outliers exist

- **Min-Max Scaling:** It converts all values into a fixed range (like 0–1).  
  - *Problem:* If there are outliers, they stretch the scale, and most of the other data points get compressed into a tiny range. 

- **Z-score Standardization:** It uses the mean and standard deviation to scale data.  
  - *Advantage:* Outliers have less effect on the overall scaling, and the data distribution is better preserved.
