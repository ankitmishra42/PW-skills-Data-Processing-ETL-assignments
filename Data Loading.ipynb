**Q1. Data Understanding: Identify all data quality issues present in the dataset that can cause problems during data loading.**

- Data Quality Issues in the Dataset:

  - **Duplicate Records:** *Order O101 with customer C001 appears twice, causing duplication.*
  - **NULL Values:** *Sales_Amount is NULL for order O102, which can break aggregations.*
  - **Invalid Data Type:** *Sales_Amount contains text "Three Thousand" instead of numeric value.*
  - **Inconsistent Date Formats:** *Order_Date uses multiple formats (DD-MM-YYYY and YYYY/MM/DD).*
  - **Incorrect Data Type Design:** *Numeric and date fields stored as text can cause loading and conversion errors.*
  - **Lack of Primary Key Constraint:** *No unique constraint on Order_ID, allowing duplicate entries.*
  - **Data Standardization Issues:** *Mixed formats reduce data reliability and complicate transformations.*

**Q7. Pre-Load Validation Checklist:**<br>
- **List the exact pre-load validation checks you would perform on this dataset before loading.**
  - We need to Check primary key uniqueness "Order_ID".
  - We need to Identify NULL values in critical columns.
  - We need to Validate numeric fields "Sales_Amount".
  - We need to Validate date formats in "Order_Date".

**Q8. Cleaning Strategy:**<br>
- **Describe the step-by-step cleaning actions required to make this dataset load-ready.**
  - We need to Remove or deduplicate repeated Order_ID records.
  - We need to Handle NULL Sales_Amount (impute, correct, or reject record).
  - We need to Convert text amounts (e.g., “Three Thousand”) to numeric or discard.
  - We need to Standardize Order_Date to one format (e.g., YYYY-MM-DD).
  - We need to Cast columns to correct data types (DECIMAL, DATE).
  - We need to Apply primary key constraint before loading.

- **a) What incorrect results might appear in Total Sales KPI?**
  - Sales total may be overstated (due to duplicates).
  - Sales may be understated (due to NULL values).
  - Some records may be ignored due to invalid data.
- **b) Which records specifically would cause misleading insights?**
  - Duplicate record: Order_ID = O101.
  - NULL amount: Order_ID = O102.
  - Invalid amount: Order_ID = O104.
- **c) Why would BI tools not detect these issues automatically?**
  - BI tools assume data is clean.
  - They perform calculations without validation.
  - Data quality checks must be done before loading, not at visualization level.
